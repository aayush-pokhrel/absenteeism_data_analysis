```python
import tensorflow as tf 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
raw_csv_data=pd.read_csv('Absenteeism_data.csv')
raw_csv_data
```

```python
df=raw_csv_data.copy()
df.info()
```

```python
df=df.drop(['ID'],axis=1)
df
```

```python
df_no_age=df.drop(['Age'],axis=1)
df_no_age
```

```python
df['Reason for Absence']
```

```python
df['Reason for Absence'].unique()
```

```python
sorted(df['Reason for Absence'].unique())
```

```python
reason_columns=pd.get_dummies(df['Reason for Absence'],dtype=int)
reason_columns
```

```python
reason_columns['check']=reason_columns.sum(axis=1)
reason_columns
```

```python
reason_columns['check'].sum(axis=0)
```

```python
reason_columns['check'].unique()
```

```python
reason_columns=reason_columns.drop(['check'],axis=1)
reason_columns
```

```python
df.columns.values
```

```python
#dropped reason for absence col to exclude the perfect multi collinearity 
df=df.drop(['Reason for Absence'],axis=1)
df
```

```python
#grouping the reason for absence column
reason_type1=reason_columns.iloc[:, 1:15].max(axis=1)
reason_type2=reason_columns.iloc[:, 15:18].max(axis=1)
reason_type3=reason_columns.iloc[:, 18:22].max(axis=1)
reason_type4=reason_columns.iloc[:, 22:].max(axis=1)
```

```python
reason_type2
```

```python
#concatenating the reason types
df=pd.concat([df,reason_type1,reason_type2,reason_type3,reason_type4],axis=1)
df
```

```python
col_names=['Date', 'Transportation Expense', 'Distance to Work', 'Age',
       'Daily Work Load Average', 'Body Mass Index', 'Education',
       'Children', 'Pets', 'Absenteeism Time in Hours', 'reason_1', 'reason_2', 'reason_3', 'reason_4']

df.columns=col_names
df.head()
```

```python
cols_reordered=['reason_1', 'reason_2', 'reason_3', 'reason_4','Date', 'Transportation Expense', 'Distance to Work', 'Age',
       'Daily Work Load Average', 'Body Mass Index', 'Education',
       'Children', 'Pets', 'Absenteeism Time in Hours']

df=df[cols_reordered]
df
```

```python
df_reason_mod=df.copy()
df_reason_mod['Date']
```

```python
df_reason_mod['Date']=pd.to_datetime(df_reason_mod['Date'], format="%d/%m/%Y")
df_reason_mod['Date']
```

```python
df_reason_mod['Date'][0]
```

```python
df_reason_mod['Date'][0].month
```

```python
#extracting monthh from date
month_list=[]
for i in range(700):
    month_list.append(df_reason_mod['Date'][i].month)

month_list
```

```python
unique_values=set(month_list)
unique_values
```

```python
df_reason_mod['Month value']=month_list
df_reason_mod
```

Extracting day of the week

```python
df_reason_mod['Date'][12].weekday()
```

```python
def date_to_weekday(date_value):
    return date_value.weekday()

df_reason_mod['Day of the week']=df_reason_mod['Date'].apply(date_to_weekday)

df_reason_mod.head()
```

```python
df_reason_mod=df_reason_mod.drop(['Date'],axis=1)
df_reason_mod
```

```python
df_reason_mod.columns
```

```python
col_list=['reason_1', 'reason_2', 'reason_3', 'reason_4','Month value', 'Day of the week',
       'Transportation Expense', 'Distance to Work', 'Age',
       'Daily Work Load Average', 'Body Mass Index', 'Education', 'Children',
       'Pets', 'Absenteeism Time in Hours', ]

df_reason_mod=df_reason_mod[col_list]
df_reason_mod
```
